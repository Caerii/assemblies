Question: which parts of the runtime affect scaling before we run a benchmark?

This notebook inspects stable engineering APIs without making hardware claims.

```python
import numpy as np
import pandas as pd

from neural_assemblies.compute import (
    RelativeThresholdPolicy,
    ThresholdPolicy,
    TopKPolicy,
    WinnerSelector,
)
from neural_assemblies.core.backend import detect_best_engine
```

```python
pd.DataFrame([
    {"n_hint": 0, "selected_engine": detect_best_engine(0)},
    {"n_hint": 100_000, "selected_engine": detect_best_engine(100_000)},
    {"n_hint": 1_000_000, "selected_engine": detect_best_engine(1_000_000)},
])
```

```python
features = np.array([0.1, 0.9, 0.5, 0.8, 0.2])
selector = WinnerSelector(np.random.default_rng(0))
policies = [
    TopKPolicy(k=2),
    ThresholdPolicy(k=3, threshold=0.5),
    RelativeThresholdPolicy(fraction_of_max=0.75, min_winners=1, max_winners=3),
]
pd.DataFrame([
    {"policy": type(policy).__name__, "selected_indices": selector.select_with_policy(features, policy).tolist()}
    for policy in policies
])
```
