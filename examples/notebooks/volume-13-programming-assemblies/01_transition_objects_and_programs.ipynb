Question: what would it mean to treat assembly operations as a small program?

This notebook starts with typed transitions because they are a stable, inspectable programming surface.

```python
import pandas as pd

from neural_assemblies.assembly_calculus import TransitionMap
```

```python
program = TransitionMap([
    ("start", "see_red", "color_bound"),
    ("color_bound", "see_triangle", "object_bound"),
    ("object_bound", "act", "done"),
])

pd.DataFrame([
    {"from_state": item.from_state, "symbol": item.symbol, "to_state": item.to_state, "probability": item.probability}
    for item in program
])
```
