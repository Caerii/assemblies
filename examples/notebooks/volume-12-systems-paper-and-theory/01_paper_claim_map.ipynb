# Volume 12. Paper Claim Map

Question: how do ambitious paper ideas become defensible sections?

This notebook reads planning notes and turns them into a claim-work checklist.

```python
from pathlib import Path

import pandas as pd
```

```python
repo_root = Path.cwd()
while not (repo_root / "research" / "plans").exists():
    if repo_root.parent == repo_root:
        raise RuntimeError("Could not locate repository root")
    repo_root = repo_root.parent

vision = repo_root / "research" / "plans" / "ASSEMBLY_SYSTEMS_PAPER_VISION.md"
sections = []
for line in vision.read_text(encoding="utf-8").splitlines():
    if line.startswith("## "):
        sections.append({"level": 2, "section": line.removeprefix("## "), "source": vision.name})
    elif line.startswith("### "):
        sections.append({"level": 3, "section": line.removeprefix("### "), "source": vision.name})

pd.DataFrame(sections).head(20)
```
