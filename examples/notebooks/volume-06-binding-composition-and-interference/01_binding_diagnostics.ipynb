# Volume 6. Binding Diagnostics

Question: after two assemblies are merged, what diagnostics tell us whether the target is useful?

This is a compact binding probe. It is not yet a capacity or interference experiment.

```python
import pandas as pd

from neural_assemblies.assembly_calculus.tracing import (
    merge_trace,
    project_trace,
    source_response_traces,
)
from neural_assemblies.core.brain import Brain
```

```python
N = 2_500
K = 45
BETA = 0.08

brain = Brain(p=0.05, save_winners=True, seed=23, engine="numpy_sparse")
for stimulus in ["red", "triangle"]:
    brain.add_stimulus(stimulus, K)
for area in ["COLOR", "SHAPE", "OBJECT"]:
    brain.add_area(area, N, K, BETA)

red = project_trace(brain, "red", "COLOR", rounds=7).final
triangle = project_trace(brain, "triangle", "SHAPE", rounds=7).final
bound = merge_trace(brain, "COLOR", "SHAPE", "OBJECT", rounds=6)

pd.DataFrame([bound.summary()])
```

```python
responses = source_response_traces(
    brain,
    ["COLOR", "SHAPE"],
    "OBJECT",
    reference=bound.final,
    labels=["COLOR alone", "SHAPE alone"],
    rounds=6,
)
pd.DataFrame(responses.to_records())
```
