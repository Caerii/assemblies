# Volume 5. Order Parameters And Regimes

Question: which numbers make assembly formation visible as a dynamical regime instead of a final snapshot?

This notebook runs a small projection sweep. It is a teaching probe, not a phase-diagram experiment.

```python
import pandas as pd

from neural_assemblies.assembly_calculus.tracing import (
    ProjectionSweepConfig,
    projection_sweep,
)
```

A projection trace gives us per-round dynamics. A sweep lets us compare small parameter changes without turning the notebook into a long experiment.

```python
configs = [
    ProjectionSweepConfig(n=1_500, k=30, beta=0.04, rounds=4, seed=11),
    ProjectionSweepConfig(n=1_500, k=30, beta=0.08, rounds=6, seed=11),
    ProjectionSweepConfig(n=2_500, k=40, beta=0.08, rounds=6, seed=12),
]
results = pd.DataFrame(projection_sweep(configs))
results
```

The table separates control parameters from order parameters. In a full experiment, these rows become many seeds and wider sweeps.

```python
pd.DataFrame([
    {"kind": "control parameter", "name": name}
    for name in ["n", "k", "density", "beta", "rounds", "seed"]
] + [
    {"kind": "order parameter", "name": name}
    for name in ["chance_overlap", "final_overlap_prev", "stabilized_at", "new_winners_final_round"]
])
```
